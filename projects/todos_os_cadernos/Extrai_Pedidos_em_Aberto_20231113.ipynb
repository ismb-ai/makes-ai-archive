```python
import pandas as pd
import requests
import json
import datetime
```

```python
# INPUTS
data = datetime.date(2023, 11, 13)
hora = '0900'

#SET UP
dia = data.day
mes = data.month
ano = data.year
s_dia = '{:02d}'.format(data.day)
s_mes = '{:02d}'.format(data.month)
s_ano = str(ano)
s_data = s_ano+ s_mes + s_dia
```

```python
url = "https://dm.makes.ai/api/get_data_db"

payload = json.dumps({
  "client": "cpp",
  "query": "consult_open_orders",
  "sync": 0
})
headers = {
  'Content-Type': 'application/json'
}

response = requests.request("POST", url, headers=headers, data=payload)

#print(response.text)
```

```python
result = json.loads(response.text)
df_result = pd.DataFrame.from_dict(result)
df_result = df_result.data.apply(pd.Series)
```

```python
# Salvando
df_result.to_csv(r'pedidos_em_aberto_'+ s_data + hora + '.csv')
```
