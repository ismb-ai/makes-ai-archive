```python
from sympy import symbols, diff, integrate, simplify, Eq, solve

# Define symbols
t, phi, phi_prime = symbols('t phi phi_prime', real=True)
L = symbols('L', real=True, positive=True)  # Lagrangian density

# Define Lagrangian density (replace this with your specific expression)
L = phi_prime**2 / 2 - (diff(phi, t))**2 / 2

# Define the action integral
action_integral = integrate(L, (t, 0, 1))

# Define the variation of the action integral
delta_phi = symbols('delta_phi', real=True)
delta_action = action_integral.subs(phi, phi + delta_phi) - action_integral

# Compute the variation of the action integral
delta_action = simplify(delta_action)

# Set the variation to zero to find the equations of motion
equation_of_motion = Eq(delta_action, 0)

# Solve the equation of motion for delta_phi
solution = solve(equation_of_motion, delta_phi)

# Print the result
print("Equations of Motion:")
for i, sol in enumerate(solution):
    print(f"Equation {i+1}: {sol}")
```
